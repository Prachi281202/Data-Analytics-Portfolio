```python
# LOADING THE REQUIRED LIBRARIES
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
# LOADING THE IPL MATCHES DATASET
ipl=pd.read_csv('ipl.csv')
```

```python
# HAVING A GLANCE AT THE FIRST 5 RECORDS OF DATA SET
ipl.head()
```

```python
#  LOOKING AT THE NUMBER OF ROWS AND COLUMNS IN THE DATASET
ipl.shape
```

```python
# GETTING THE FREQUENCY OF MOST MAN 0F THE MATCH AWARDS
ipl['player_of_match'].value_counts()
```

```python
# GETTING THE TOP 10 PLAYERS WITH MOST PLAYER OF THE MATCH AWARDS
ipl['player_of_match'].value_counts()[0:10]
```

```python
# GETTING THE TOP 5 PLAYERS WITH MOST PLAYER OF THE MATCH AWARDS
ipl['player_of_match'].value_counts()[0:5]
```

```python
list(ipl['player_of_match'].value_counts()[0:5].keys())
```

```python
plt.figure(figsize=(8,5))
plt.bar(list(ipl['player_of_match'].value_counts()[0:5].keys()),list(ipl['player_of_match'].value_counts()[0:5]),color="g")
plt.show()
```

```python
# MAKING A BAR-PLOT FOR THE TOP 5 PLAYERS WITH MOST MAN OF THE MATCH AWARDS
plt.figure(figsize=(8,5))
plt.bar(list(ipl['player_of_match'].value_counts()[0:5].keys()),list(ipl['player_of_match'].value_counts()[0:5]),color="black")
plt.show()
```

```python
# GETTING THE FREQUENCY OF RESULT COLUMN
ipl['result'].value_counts()
```

```python
# FINDING OUT THE NUMBER OF TOSS WINS W.R.T EACH TEAMS
ipl['toss_winner'].value_counts()
```

```python
# EXTRACTING THE RRECORDS WHERE A TEAM WON BATTING FIRST
batting_first=ipl[ipl['win_by_runs']!=0]
```

```python
# LOOKING AT THE HEAD
batting_first.head()
```

```python
# MAKING A HISTOGRAM
plt.figure(figsize=(5,5))
plt.hist(batting_first['win_by_runs'])
plt.title("Distribution of runs")
plt.xlabel("Runs")
plt.show()
```

```python
#  FINDING OUT THE NUMBER OF WINS W.R.T EACH TEAM AFTER BATTING FIRST
batting_first['winner'].value_counts()
```

```python
# MAKING A BAR-PLOT FOR TOP 5 TEAMS WITH MOST WINS AFTER BATTING FIRST
plt.figure(figsize=(11,4))
plt.bar(list(batting_first['winner'].value_counts()[0:5].keys()),list(batting_first['winner'].value_counts()[0:5]),color=["black"])
plt.show()
```

```python
# MAKING A PIE CHART
plt.figure(figsize=(8,11))
plt.pie(list(batting_first['winner'].value_counts()),labels=list(batting_first['winner'].value_counts().keys()),autopct='%0.1f%%')
plt.show()
```

```python
#  EXTRACTING THOSE RECORDS WHERE A TEAM HAS WON AFTER BATTING SECOND
batting_second=ipl[ipl['win_by_wickets']!=0]
```

```python
# LOOKING AT THE HEAD
batting_second.head()
```

```python
# MAKING A HISTOGRAM
plt.figure(figsize=(10,6))
plt.hist(batting_second['win_by_wickets'],bins=30)
plt.show()
```

```python
# FINDING OUT THE FREQUENCY OF NUMBER OF WINS W.R.T EACH TIME AFTER BATTING SECOND
batting_second['winner'].value_counts()
```

```python
# MAKING A BAR-PLOT FOR TOP 5 TEAMS WITH MOST WINS AFTER BATTING SECOND
plt.figure(figsize=(11,4))
plt.bar(list(batting_second['winner'].value_counts()[0:5].keys()),list(batting_second['winner'].value_counts()[0:5]),color=["black","red","green","pink","blue"])
plt.show()
```

```python
# MAKING A PIE CHART FOR DISTRIBUTION OF MOST WINS AFTER BATTING SECOND
plt.figure(figsize=(8,11))
plt.pie(list(batting_second['winner'].value_counts()),labels=list(batting_second['winner'].value_counts().keys()),autopct='%0.1f%%')
plt.show()
```

```python
#  LOOKING AT THE NUMBER OF MATCHES PLAYED IN EACH CITY
ipl['city'].value_counts()
```
